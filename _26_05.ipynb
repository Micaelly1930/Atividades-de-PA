```python
# projeto caixa eletronico

# voce foi contratado para desenvolver um sistema simples de caixa eletronico

#  o sistema devera duncionar atraves de um menu interativo utilizando "while"

# o progrma deve iniciar com m saldo inicial de R$:1000,00 e permitir que o usuario escolhe diferentes operações ate decidir sair do sistema .

#ver saldo
# depositar
# sacar dinheiro
# sair

# regras:

# quando depositar soma ao saldo anterior:
# quando sacar subtrair so salto anterior
# mostrar saldo para usuario
```

```python
opcao = ""
saldo = 1000.00
print ("==========  MENU =========")
print("")
print("opção 1 - ver saldo")
print("opção 2 - depositar")
print("opção 3 - sacar dinheiro")
print("opção 4 - sair")
print("==============================")
print("")

while opcao != 4:
  try:
    opcao = int(input(" escolha uma opção que deseja ou digite 4 para sair : "))

    if opcao == 1:
      print(f"Seu saldo atual é: R$ {saldo:}")
      print("")

    elif opcao == 2:
      valor_deposito = float(input("Digite o valor a depositar: "))
      if valor_deposito > 0:
        saldo += valor_deposito
        print(f"Depósito de R$ {valor_deposito:} realizado com sucesso.")
        print(f"Seu novo saldo é: R$ {saldo:}")
      else:
        print("Valor de depósito inválido. Digite um valor positivo.")
      print("")

    elif opcao == 3:
      valor_saque = float(input("Digite o valor a sacar: "))
      if valor_saque > 0 and valor_saque <= saldo:
        saldo -= valor_saque
        print(f"Saque de R$ {valor_saque:} realizado com sucesso.")
        print(f"Seu novo saldo é: R$ {saldo:}")
      elif valor_saque > saldo:
        print("Saldo insuficiente para realizar o saque.")
      else:
        print("Valor de saque inválido. Digite um valor positivo.")
      print("")

    elif opcao == 4:
      print("Saindo do sistema. Obrigado!")
      print("")

    else:
      print("Opção inválida. Por favor, escolha uma opção entre 1 e 4.")
      print("")

  except ValueError:
    print("Entrada inválida. Por favor, digite um número para a opção.")
    print("")
```

```python
#azer um jogo de adivinhação
#O usuario ira adivinhar um numero de 0 a 10.
#Regras
#O usuario tera 03 chences para adivinhar
#informar se o palpite do usuario e maior ou menor que o numero
#imprimir mensagen de está quente, morno ou frio

import random

a = random.randint(0, 10)

chances = 3
numero = -1
print("Bem-vindo ao jogo de adivinhação!")
print(f"Você tem {chances} chances para adivinhar um número entre 0 e 10.")

while numero != a and chances > 0:
    try:
        numero = float(input(f"(Chance {4 - chances}/3) Digite um numero de 0 a 10: "))

        if 0 <= numero <= 10:
            if numero < a :
                print("O número é maior")
                chances -= 1
            elif numero > a :
                print("O número é menor")
                chances -= 1
            elif numero == a:
                print("Parabéns! Você acertou!")
                break ,
        else:
            print("Por favor, digite um número entre 0 e 10.")

    except ValueError:
        print("Entrada inválida. Por favor, digite um número.")

```

```python
print("=== Simulador de Saque ===")
valor_saque = 0

while True:
    try:
        valor_saque_str = input("Digite o valor que deseja sacar (apenas números inteiros, ou 'sair' para encerrar): ")
        if valor_saque_str.lower() == 'sair':
            print("Encerrando o simulador de saque.")
            break

        valor_saque = int(valor_saque_str)

        if valor_saque <= 0:
            print("Por favor, digite um valor positivo para o saque.")
            continue

        # Define the available notes (in descending order for easier calculation)
        notas = [100, 50, 20, 10, 5, 2, 1]
        notas_entregues = {}
        valor_restante = valor_saque

        print(f"\nPara sacar R$ {valor_saque:.2f}, você receberá:")

        for nota in notas:
            if valor_restante >= nota:
                quantidade = valor_restante // nota
                if quantidade > 0:
                    notas_entregues[nota] = quantidade
                    valor_restante -= quantidade * nota

        if not notas_entregues:
            print("Não foi possível entregar nenhuma nota para este valor (valor muito baixo ou notas indisponíveis).")
        else:
            for nota, quantidade in notas_entregues.items():
                print(f"  {quantidade} nota(s) de R$ {nota:.2f}")

        if valor_restante > 0:
            print(f"Sobrou R$ {valor_restante:.2f} que não pode ser distribuído em notas disponíveis (apenas notas inteiras são consideradas). Você pode tentar sacar este valor em uma próxima operação ou em outra modalidade.")
        print("--------------------------------------")

    except ValueError:
        print("Entrada inválida. Por favor, digite um número inteiro para o valor do saque ou 'sair'.")
    except Exception as e:
        print(f"Ocorreu um erro inesperado: {e}")
```
